# 1-1. 대표 통계량

    - 평균, 조화평균, 중앙값, 최빈값 등 데이터의 대표값을 Python으로 계산해본다

```python
import numpy as np  # 수치 계산용 NumPy를 np라는 짧은 이름으로 불러옴
import pandas as pd  # 표 형태 데이터 처리용 Pandas를 pd라는 짧은 이름으로 불러옴
from scipy import stats  # 통계 함수가 있는 scipy.stats 모듈을 불러옴
```

# 산술 평균

```python
x = [1, 2, 3, 4, 5]  # 평균을 구할 숫자 목록

print(np.mean(x))  # NumPy 함수로 평균 계산
print(np.array(x).mean())  # 리스트를 NumPy 배열로 바꾼 뒤 평균 계산
print(pd.Series(x).mean())  # 리스트를 Pandas Series로 바꾼 뒤 평균 계산
```

# 조화 평균

```python
speed = [60, 40]  # 같은 거리를 이동한 두 구간의 속도

print(np.mean(speed))  # 산술평균: (60 + 40) / 2
print(stats.hmean(speed))  # 조화평균: 같은 거리 이동의 평균 속도에 적합

print(round(stats.hmean(speed), 2))  # 소수점 둘째 자리까지 반올림
print(f"{stats.hmean(speed):.2f}")  # 소수점 둘째 자리까지 형식을 맞춰 출력
```

## 같은 거리를 이동했을 때 평균 속도

```python
distance = 60  # 각 구간에서 이동한 거리(km)

total_distance = distance * 2  # 전체 이동 거리: 60km + 60km
total_time = distance / 60 + distance / 40  # 각 구간의 이동 시간을 더함

average_speed = total_distance / total_time  # 전체 거리 ÷ 전체 시간

print(f"실제 평균 속도: {average_speed:.2f}km/h")  # 실제 계산 결과
print(f"조화평균: {stats.hmean(speed):.2f}km/h")  # 조화평균 결과와 비교
```

## 절사평균

양끝의 극단값 일부를 제외한 뒤 평균을 계산한다.

```python
income = [280, 290, 300, 310, 320, 330, 340, 350, 360, 370, 3000]  # 소득 데이터

print(f"산술평균: {np.mean(income):.2f}")  # 모든 값을 포함해 평균 계산
print(f"절사평균: {stats.trim_mean(income, 0.1):.2f}")  # 양끝 10%씩 제외해 평균 계산
```

최빈값

## Pandas 최빈값

```python
x = [1, 2, 2, 3, 3, 3, 4, 5]  # 최빈값을 찾을 데이터

count = pd.Series(x).value_counts()  # 값별 등장 횟수를 많은 순서대로 계산

print(count)  # 값과 등장 횟수 전체 출력
print(count.index[0])  # 가장 많이 등장한 첫 번째 값, 즉 최빈값 출력
```

## Counter 최빈값

```python
from collections import Counter  # 값별 등장 횟수를 세는 파이썬 기본 도구

counter = Counter(x)  # x의 각 값이 나온 횟수를 세어 저장

print(counter.most_common(1))  # 가장 많이 나온 값 1개를 [(값, 횟수)] 형태로 출력
print(counter.most_common(1)[0])  # 첫 번째 결과를 (값, 횟수) 형태로 출력
print(counter.most_common(1)[0][0])  # (값, 횟수) 중 값만 꺼내 최빈값 출력
```

# 중앙값

```python
x = [1, 2, 3, 4, 100]  # 100이라는 극단값이 포함된 데이터

print(f"산술평균: {np.mean(x):.1f}")  # 모든 값의 평균
print(f"중앙값: {np.median(x):.1f}")  # 정렬했을 때 가운데 값
```

## NumPy | SciPy로 대표 통계량 한 번에 계산

```python
data = np.array([1, 2, 2, 3, 3, 3, 4, 5])  # 통계 계산에 사용할 NumPy 배열

mean = np.mean(data)  # 평균 계산
median = np.median(data)  # 중앙값 계산
mode = stats.mode(data)  # 최빈값과 최빈값 빈도를 함께 계산

print(f"평균: {mean}")  # 계산한 평균 출력
print(f"중앙값: {median}")  # 계산한 중앙값 출력
print(f"최빈값: {mode.mode}")  # mode 결과 중 최빈값 출력
print(f"최빈값 빈도: {mode.count}")  # 최빈값이 나온 횟수 출력
```

## Pandas로 대표 통계량 한 번에 계산

```python
series = pd.Series([1, 2, 2, 3, 3, 3, 4, 5])  # 리스트를 Pandas Series로 변환

print(f"평균: {series.mean()}")  # Series의 평균 계산
print(f"중앙값: {series.median()}")  # Series의 중앙값 계산
print(f"최빈값: {series.mode().to_list()}")  # 최빈값을 리스트 형태로 변환해 출력
```
