# 1-1. 대표 통계량

데이터의 중심을 하나의 값으로 요약하는 대표 통계량을 계산한다.

- 산술평균, 조화평균, 절사평균, 중앙값, 최빈값의 정의와 차이를 확인한다.
- 데이터의 성격과 이상치 유무에 따라 적절한 대표값을 선택한다.

```python
# 셀 목적: 대표 통계량 계산에 필요한 NumPy·Pandas·SciPy를 불러온다.
# 해석 포인트: 같은 대표값도 배열·Series·통계 전용 함수 등 여러 방식으로 계산할 수 있다.
import numpy as np  # 산술평균·중앙값과 배열 연산을 위해 NumPy를 np로 불러옴
import pandas as pd  # Series의 평균·중앙값·최빈값 계산을 위해 Pandas를 pd로 불러옴
from scipy import stats  # 조화평균·절사평균·최빈값 계산을 위해 scipy.stats를 불러옴
```

## 1. 산술평균

산술평균은 모든 값을 더한 뒤 자료 수로 나눈 값이다.
전체 자료의 중심을 반영하지만, 매우 크거나 작은 이상치의 영향을 크게 받는다.

```python
# 셀 목적: 동일한 데이터의 산술평균을 NumPy와 Pandas의 세 가지 방식으로 계산한다.
# 해석 포인트: 세 방법은 모두 (1+2+3+4+5)/5를 계산하므로 결과가 3으로 일치해야 한다.
x = [1, 2, 3, 4, 5]  # 산술평균 계산에 사용할 다섯 개의 관측값을 리스트로 저장

print(np.mean(x))  # NumPy의 mean()으로 계산한 x의 산술평균 3을 출력
print(np.array(x).mean())  # x를 NumPy 배열로 변환한 뒤 배열 메서드로 계산한 산술평균 3을 출력
print(pd.Series(x).mean())  # x를 Pandas Series로 변환한 뒤 계산한 산술평균 3을 출력
```

## 2. 조화평균

조화평균은 각 값의 역수를 산술평균한 뒤 다시 역수를 취한 값이다.
같은 거리에서 측정한 속도처럼 분모가 되는 비율이나 단위당 값의 평균에 적합하다.

```python
# 셀 목적: 두 구간 속도의 산술평균과 조화평균을 계산해 비율 자료의 평균 차이를 비교한다.
# 해석 포인트: 같은 거리를 60km/h와 40km/h로 이동하면 느린 구간에 더 오래 머물기 때문에 실제 평균속도는 조화평균 48km/h이다.
speed = [60, 40]  # 같은 거리를 이동한 두 구간의 속도를 km/h 단위로 저장

print(np.mean(speed))  # 두 속도를 단순히 더해 2로 나눈 산술평균 50km/h를 출력
print(stats.hmean(speed))  # 속도의 역수에 가중되는 조화평균 48km/h를 출력

print(round(stats.hmean(speed), 2))  # 조화평균을 실제 숫자 48.0으로 반올림해 출력
print(f"{stats.hmean(speed):.2f}")  # 조화평균을 소수 둘째 자리까지 표시한 문자열 48.00으로 출력
```

### 2-1. 같은 거리를 이동했을 때 평균속도 검산

평균속도의 정의인 `전체 거리 ÷ 전체 시간`으로 직접 계산해 조화평균 결과와 비교한다.

```python
# 셀 목적: 전체 거리와 전체 시간을 이용해 실제 평균속도를 직접 계산하고 조화평균과 비교한다.
# 해석 포인트: 정의로 계산한 평균속도와 조화평균이 모두 48km/h이면 같은 거리의 속도 평균에 조화평균이 적합함을 확인할 수 있다.
distance = 60  # 두 구간에서 각각 이동한 거리를 60km로 설정

total_distance = distance * 2  # 두 구간의 이동거리 60km를 더해 전체 거리 120km를 계산
total_time = distance / 60 + distance / 40  # 각 구간의 거리/속도를 더해 전체 이동시간 2.5시간을 계산

average_speed = total_distance / total_time  # 전체 거리 120km를 전체 시간 2.5시간으로 나눠 실제 평균속도를 계산

print(f"실제 평균 속도: {average_speed:.2f}km/h")  # 전체 거리와 시간으로 직접 계산한 실제 평균속도 48.00km/h를 출력
print(f"조화평균: {stats.hmean(speed):.2f}km/h")  # SciPy 조화평균 48.00km/h를 출력해 직접 계산값과 비교
```

## 3. 절사평균

절사평균은 정렬한 데이터의 양쪽 끝에서 일정 비율을 제거한 뒤 계산한 평균이다.
극단값의 영향을 줄이면서도 중앙 부근의 여러 관측값을 함께 반영하고 싶을 때 사용한다.

```python
# 셀 목적: 극단적으로 큰 소득값을 포함한 자료에서 산술평균과 10% 절사평균을 비교한다.
# 해석 포인트: 산술평균은 3000의 영향을 크게 받지만, 절사평균은 양끝 값을 제거해 일반적인 소득 수준에 가까운 중심을 보여준다.
income = [280, 290, 300, 310, 320, 330, 340, 350, 360, 370, 3000]  # 대부분 280~370이지만 극단값 3000을 포함한 소득 자료를 저장

print(f"산술평균: {np.mean(income):.2f}")  # 극단값 3000을 포함한 전체 소득의 산술평균을 출력
print(f"절사평균: {stats.trim_mean(income, 0.1):.2f}")  # 정렬 후 양끝에서 10%씩 제거해 계산한 절사평균을 출력
```

## 4. 최빈값

최빈값은 데이터에서 가장 자주 나타나는 값이다.
수치형 자료뿐 아니라 혈액형·선호 브랜드 같은 범주형 자료의 대표값으로도 사용할 수 있다.

### 4-1. Pandas로 최빈값 계산

`value_counts()`로 각 값의 빈도를 센 뒤 가장 높은 빈도의 값을 선택한다.

```python
# 셀 목적: Pandas의 value_counts()로 값별 빈도표를 만들고 최빈값을 찾는다.
# 해석 포인트: 값 3이 세 번으로 가장 많이 나타나므로 빈도표의 첫 번째 인덱스이자 최빈값은 3이다.
x = [1, 2, 2, 3, 3, 3, 4, 5]  # 값 3이 가장 자주 나타나도록 구성한 최빈값 실습 자료를 저장

count = pd.Series(x).value_counts()  # x를 Series로 변환하고 각 값의 등장 횟수를 내림차순으로 계산

print(count)  # 각 값과 등장 횟수를 빈도가 높은 순서로 정리한 전체 빈도표를 출력
print(count.index[0])  # 빈도표 첫 행의 인덱스인 최빈값 3을 출력
```

### 4-2. Counter로 최빈값 계산

파이썬 표준 라이브러리의 `Counter`를 사용해 값과 빈도를 단계적으로 추출한다.

```python
# 셀 목적: Counter로 값별 등장 횟수를 세고 most_common() 결과에서 최빈값을 단계적으로 추출한다.
# 해석 포인트: most_common(1)은 가장 빈도가 높은 한 항목을 반환하며, 결과의 첫 번째 튜플에서 값 3과 빈도 3을 구분해 꺼낼 수 있다.
from collections import Counter  # 값별 등장 횟수를 세는 파이썬 표준 라이브러리 Counter를 불러옴

counter = Counter(x)  # 앞에서 만든 x의 각 값과 등장 횟수를 딕셔너리 형태의 Counter 객체에 저장

print(counter.most_common(1))  # 최빈 항목 하나를 포함한 리스트 [(3, 3)]을 출력
print(counter.most_common(1)[0])  # 최빈값과 빈도로 구성된 첫 번째 튜플 (3, 3)을 출력
print(counter.most_common(1)[0][0])  # 첫 번째 튜플의 값 부분만 선택해 최빈값 3을 출력
```

## 5. 중앙값

중앙값은 데이터를 크기순으로 정렬했을 때 가운데 위치하는 값이다.
극단값의 크기보다 순서만 사용하므로 이상치나 치우친 분포의 중심을 나타낼 때 산술평균보다 안정적이다.

```python
# 셀 목적: 극단값이 포함된 데이터의 산술평균과 중앙값을 비교한다.
# 해석 포인트: 극단값 100 때문에 평균은 22로 커지지만 중앙값은 가운데 관측값 3을 유지한다.
x = [1, 2, 3, 4, 100]  # 일반 값 1~4와 극단적으로 큰 값 100을 포함한 자료를 저장

print(f"산술평균: {np.mean(x):.1f}")  # 극단값 100의 영향을 받은 산술평균 22.0을 출력
print(f"중앙값: {np.median(x):.1f}")  # 정렬된 자료의 가운데 값으로 이상치에 강건한 중앙값 3.0을 출력
```

## 6. 라이브러리별 대표 통계량 계산

### 6-1. NumPy와 SciPy 사용

하나의 배열에서 평균·중앙값·최빈값과 최빈값 빈도를 함께 계산한다.

```python
# 셀 목적: NumPy와 SciPy로 동일한 배열의 평균·중앙값·최빈값·최빈값 빈도를 계산한다.
# 해석 포인트: 평균은 2.875, 중앙값과 최빈값은 3이며 값 3은 세 번 나타난다.
data = np.array([1, 2, 2, 3, 3, 3, 4, 5])  # 세 종류의 대표값을 비교할 여덟 개 관측값을 NumPy 배열로 저장

mean = np.mean(data)  # 모든 관측값의 합을 자료 수 8로 나눠 산술평균을 계산
median = np.median(data)  # 정렬된 자료의 가운데 두 값의 평균으로 중앙값을 계산
mode = stats.mode(data)  # SciPy로 최빈값과 그 값의 등장 횟수를 함께 계산

print(f"평균: {mean}")  # data의 산술평균 2.875를 출력
print(f"중앙값: {median}")  # data의 중앙값 3.0을 출력
print(f"최빈값: {mode.mode}")  # SciPy ModeResult에서 가장 자주 나온 값 3을 출력
print(f"최빈값 빈도: {mode.count}")  # SciPy ModeResult에서 최빈값 3의 등장 횟수 3을 출력
```

### 6-2. Pandas 사용

동일한 데이터를 Series로 만들어 평균·중앙값·최빈값을 계산한다.
Pandas의 `mode()`는 최빈값이 여러 개인 경우를 지원하므로 Series를 반환한다.

```python
# 셀 목적: Pandas Series 메서드로 동일한 데이터의 평균·중앙값·최빈값을 계산한다.
# 해석 포인트: NumPy·SciPy 결과와 평균 2.875, 중앙값 3.0, 최빈값 3이 일치하는지 확인한다.
series = pd.Series([1, 2, 2, 3, 3, 3, 4, 5])  # 대표값을 계산할 여덟 개 관측값을 Pandas Series로 저장

print(f"평균: {series.mean()}")  # Series의 산술평균 2.875를 출력
print(f"중앙값: {series.median()}")  # Series의 중앙값 3.0을 출력
print(f"최빈값: {series.mode().to_list()}")  # Series가 반환한 최빈값을 일반 리스트 [3]으로 변환해 출력
```
